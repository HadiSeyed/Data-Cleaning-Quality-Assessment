```python
import pandas as pd

df = pd.read_excel(
    "Film_Awards_Completed.xlsx",
    sheet_name="Tidy Data"
)

print(df.columns)
```

```python
# Example corrections
df["Winner"] = df["Winner"].str.strip()
df["Film"] = df["Film"].str.strip()

# Correct Olivia Colman's misspelling
df["Winner"] = df["Winner"].replace(
    "Olivia Coleman",
    "Olivia Colman"
)

# Make sure film titles are strings
df["Film"] = df["Film"].astype("string")

# Specifically preserve 1917 as a film title
df["Film"] = df["Film"].replace(
    1917,
    "1917"
)

# Remove unwanted whitespace
df["Winner"] = df["Winner"].str.replace(
    r"\s+",
    " ",
    regex=True
)

df["Film"] = df["Film"].str.replace(
    r"\s+",
    " ",
    regex=True
)

print(df)
```
