```python
import pandas as pd


INPUT_FILE = "Film_Awards.xlsx"


def clean_text(value):
    """Clean whitespace and convert values to strings."""
    if pd.isna(value):
        return None

    value = str(value).strip()

    if value == "":
        return None

    return value


def clean_excel_data(filename):
    # Read the Excel workbook without assuming the first row
    # is necessarily the header.
    raw = pd.read_excel(filename, header=None)

    # Remove completely empty rows and columns.
    raw = raw.dropna(axis=0, how="all")
    raw = raw.dropna(axis=1, how="all")

    # Display the raw structure so that the spreadsheet layout
    # can be inspected.
    print("Raw spreadsheet:")
    print(raw)

    return raw


if __name__ == "__main__":
    df = clean_excel_data(INPUT_FILE)
```

```python
# Step 2: Create the tidy DataFrame
import pandas as pd

tidy_rows = []

# Best Picture
for row in range(2, 5):
    year = df.iloc[row, 0]

    # Academy Award
    tidy_rows.append({
        "Year": year,
        "Award Category": "Best Picture",
        "Organization": "Academy Award",
        "Winner": None,
        "Film": df.iloc[row, 1],
        "Director": df.iloc[row, 2]
    })

    # Golden Globe - Drama
    tidy_rows.append({
        "Year": year,
        "Award Category": "Best Picture",
        "Organization": "Golden Globe - Drama",
        "Winner": None,
        "Film": df.iloc[row, 3],
        "Director": df.iloc[row, 4]
    })

    # Golden Globe - Musical or Comedy
    tidy_rows.append({
        "Year": year,
        "Award Category": "Best Picture",
        "Organization": "Golden Globe - Musical or Comedy",
        "Winner": None,
        "Film": df.iloc[row, 5],
        "Director": df.iloc[row, 6]
    })


# Best Actor
for row in range(9, 12):
    year = df.iloc[row, 0]

    # Academy Award
    tidy_rows.append({
        "Year": year,
        "Award Category": "Best Actor",
        "Organization": "Academy Award",
        "Winner": df.iloc[row, 1],
        "Film": df.iloc[row, 2],
        "Director": None
    })

    # Golden Globe - Drama
    tidy_rows.append({
        "Year": year,
        "Award Category": "Best Actor",
        "Organization": "Golden Globe - Drama",
        "Winner": df.iloc[row, 3],
        "Film": df.iloc[row, 4],
        "Director": None
    })

    # Golden Globe - Musical or Comedy
    tidy_rows.append({
        "Year": year,
        "Award Category": "Best Actor",
        "Organization": "Golden Globe - Musical or Comedy",
        "Winner": df.iloc[row, 5],
        "Film": df.iloc[row, 6],
        "Director": None
    })


# Best Actress
for row in range(16, 19):
    year = df.iloc[row, 0]

    # Academy Award
    tidy_rows.append({
        "Year": year,
        "Award Category": "Best Actress",
        "Organization": "Academy Award",
        "Winner": df.iloc[row, 1],
        "Film": df.iloc[row, 2],
        "Director": None
    })

    # Golden Globe - Drama
    tidy_rows.append({
        "Year": year,
        "Award Category": "Best Actress",
        "Organization": "Golden Globe - Drama",
        "Winner": df.iloc[row, 3],
        "Film": df.iloc[row, 4],
        "Director": None
    })

    # Golden Globe - Musical or Comedy
    tidy_rows.append({
        "Year": year,
        "Award Category": "Best Actress",
        "Organization": "Golden Globe - Musical or Comedy",
        "Winner": df.iloc[row, 5],
        "Film": df.iloc[row, 6],
        "Director": None
    })


# Convert to DataFrame
tidy_df = pd.DataFrame(tidy_rows)

display(tidy_df)
```

```python
# Step 3: Clean the data
# Remove unnecessary whitespace
for column in ["Winner", "Film", "Director"]:
    tidy_df[column] = tidy_df[column].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

# Correct known spelling errors
tidy_df["Winner"] = tidy_df["Winner"].replace(
    "Olivia Coleman",
    "Olivia Colman"
)

tidy_df["Director"] = tidy_df["Director"].replace(
    "Peter Farelly",
    "Peter Farrelly"
)

# Make sure film titles are strings
tidy_df["Film"] = tidy_df["Film"].apply(
    lambda x: str(x) if pd.notna(x) else x
)

display(tidy_df)
```

```python
# Sort the tidy data
tidy_df = tidy_df.sort_values(
    [
        "Year",
        "Award Category",
        "Organization"
    ]
).reset_index(drop=True)



# Sort the tidy data from 2019 to 2017
tidy_df = tidy_df.sort_values(
    "Year",
    ascending=False
).reset_index(drop=True)

# # Sort the tidy data from 2017 to 2019
# tidy_df = tidy_df.sort_values(
#     "Year",
#     ascending=True
# ).reset_index(drop=True)



# Display the tidy DataFrame
print("\nTIDY DATA")
print("=" * 80)

print(tidy_df.to_string(index=False))


# Save the tidy data to a new Excel file
output_file = "Film_Awards_Tidy2.xlsx"

tidy_df.to_excel(
    output_file,
    index=False,
    sheet_name="Tidy Data"
)

print("\n" + "=" * 80)
print(f"Tidy data saved to: {output_file}")
print(f"Number of rows: {len(tidy_df)}")
print(f"Number of columns: {len(tidy_df.columns)}")
```

### Question 1: For each film involved in multiple awards, list the award and year it is associated with.

```python
film_counts = (
    tidy_df
    .dropna(subset=["Film"])
    .groupby("Film")
    .size()
)

multiple_films = film_counts[film_counts > 1]

film_results = tidy_df[
    tidy_df["Film"].isin(multiple_films.index)
].sort_values(
    ["Film", "Year"]
)

display(
    film_results[
        ["Film", "Year", "Award Category", "Organization"]
    ]
)
```

### Question 2: For each actor/actress winning multiple awards, list the film and award they are associated with.

```python
winner_counts = (
    tidy_df
    .dropna(subset=["Winner"])
    .groupby("Winner")
    .size()
)

multiple_winners = winner_counts[
    winner_counts > 1
]

actor_results = tidy_df[
    tidy_df["Winner"].isin(multiple_winners.index)
].sort_values(
    ["Winner", "Year"]
)

display(
    actor_results[
        [
            "Winner",
            "Film",
            "Year",
            "Award Category",
            "Organization"
        ]
    ]
)
```
