### Description (short documentation)

### Contents
- Excel/tidy_excel.py: transforms Film_Awards.xlsx into tidy data.
- Excel/Film_Awards_Tidy.xlsx: Excel tidy data and answers.
- Excel/Film_Awards_Tidy2.xlsx: Excel tidy data and answers, sorted by year (2019 to 2017).
- SQL/tidy_sql.py: transforms the SQLite data for 2017-2019.
- SQL/Film_Awards_SQL_Results.xlsx: SQL tidy data and answers.
- Extra_Credit/create_missing.py: creates artificial missing data.
- Extra_Credit/Students.csv: created a CSV file.
- Extra_Credit/Students_na.csv: output of missing data.

### Excel
The original workbook contains three sections: Best Picture, Best Actor, and Best Actress. The script reshapes them into one tidy table with one award per row and variables represented by columns. The data are sorted from 2019 to 2017.

Known data-quality corrections:
- 1917 is retained as the film title.
- Peter Farelly is corrected to Peter Farrelly.
- Olivia Coleman is corrected to Olivia Colman.
- Leading/trailing whitespace and the newline in The Favourite are removed.

Run:
python tidy_excel.py

### SQL
The SQLite database contains award, actor, director, and organization tables. The analysis is restricted to 2017, 2018, and 2019 as required. Picture awards are related to films/directors through the director table; acting awards are related to films through the actor table.

Run:
python tidy_sql.py

### Extra Credit
Basic:
python create_missing.py Students.csv

25 percent:
python create_missing.py Students.csv --percent 25

Column 3:
python create_missing.py Students.csv --percent 20 --column 3

Custom output:
python create_missing.py Students.csv --percent 10 --output Students_na.csv

The default percentage is 15. The percentage must be between 1 and 99. Existing missing values are retained.

I created “create_missing.py” file that artificially introduces missing values into a CSV file. Then made a “Students.csv” file. By default, the program adds missing values to approximately 15% of the rows and saves the result as “Students_na.csv” file. If the column option is not provided, missing values are randomly distributed across all columns.
